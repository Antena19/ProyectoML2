# 7. Modelos de Regresión

## Evaluación 2 - Machine Learning

Este notebook implementa y evalúa 6 modelos de regresión con GridSearchCV y CrossValidation (k=5).

### Modelos:
1. Linear Regression (Ridge)
2. Random Forest Regressor
3. Gradient Boosting Regressor
4. SVR
5. KNN Regressor
6. Decision Tree Regressor

### Métricas: MAE, MSE, RMSE, R², MAPE (mean±std)


```python
# Importar librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-darkgrid')
%matplotlib inline

# Cargar desde Kedro
from kedro.framework.session import KedroSession
from pathlib import Path

project_path = Path.cwd().parent.parent
session = KedroSession.create(project_path=project_path)
catalog = session.load_context().catalog

# Cargar resultados
resultados = catalog.load("resultados_regresion")
tabla_comparativa = catalog.load("tabla_comparativa_regresion")

print(f"✓ Datos cargados - Problemas: {list(resultados.keys())}")
```

## Tabla Comparativa (mean±std)


```python
# Mostrar tabla comparativa
print("=" * 100)
print("TABLA COMPARATIVA - 6 MODELOS DE REGRESIÓN")
print("=" * 100)
display(tabla_comparativa)

tabla_comparativa.to_csv("tabla_regresion_final.csv", index=False)
print("\n✓ Tabla exportada")
```

## Visualizaciones


```python
# Gráfico de R² Score
tabla_comparativa['R2_Val'] = tabla_comparativa['R² (Test)'].astype(float)

fig, ax = plt.subplots(figsize=(14, 6))
problemas = tabla_comparativa['Problema'].unique()

for i, problema in enumerate(problemas):
    datos = tabla_comparativa[tabla_comparativa['Problema'] == problema]
    x = np.arange(len(datos))
    ax.bar(x + i*0.35, datos['R2_Val'], width=0.35, label=problema)

ax.set_xlabel('Modelo', fontweight='bold')
ax.set_ylabel('R² Score', fontweight='bold')
ax.set_title('Comparación de R² - Modelos de Regresión', fontsize=14, fontweight='bold')
ax.set_xticks(np.arange(6))
ax.set_xticklabels(datos['Modelo'], rotation=45, ha='right')
ax.legend()
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('comparacion_r2_regresion.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Gráfico guardado")
```

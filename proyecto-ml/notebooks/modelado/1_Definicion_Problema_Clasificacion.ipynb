## 🎯 Objetivo del Problema

### **PROBLEMA PRINCIPAL:**
**Predecir el SEXO de una persona fallecida** basándose únicamente en características de su muerte.

### **DATOS DISPONIBLES:**
- **Edad de fallecimiento** (0-118 años)
- **Fecha de muerte** (año, mes, día, trimestre)
- **Región geográfica** donde ocurrió la muerte
- **Features temporales** (estación, fin de semana, etc.)

### **VARIABLE OBJETIVO:**
- **SEXO**: Clasificación binaria (Hombre/Mujer)

---

## 🧠 Hipótesis Epidemiológica

Basándome en estadísticas reales de mortalidad, mi hipótesis es que:

### **1. Diferencias por Edad:**
- **HOMBRES** tienden a morir más jóvenes que mujeres
- **MUJERES** tienden a morir más mayores que hombres
- **Edad promedio de muerte** es diferente entre géneros

### **2. Patrones Temporales:**
- **Estacionalidad** puede diferir por género
- **Fin de semana vs días laborales** pueden mostrar patrones distintos
- **Meses específicos** pueden tener mayor mortalidad por género

### **3. Patrones Regionales:**
- **Diferentes regiones** pueden tener patrones de mortalidad distintos
- **Zonas geográficas** (Norte/Centro/Sur) pueden influir

---

## 📊 Ejemplos Concretos

### **Caso 1: Muerte a los 25 años**
- **Hipótesis**: Más probable que sea **HOMBRE**
- **Razón**: Accidentes, violencia, comportamientos de riesgo más comunes en hombres jóvenes

### **Caso 2: Muerte a los 80 años**
- **Hipótesis**: Más probable que sea **MUJER**
- **Razón**: Las mujeres tienen mayor esperanza de vida

### **Caso 3: Muerte en invierno**
- **Hipótesis**: Puede haber diferencias por género
- **Razón**: Diferentes patrones de enfermedades respiratorias

---

## 🎯 Objetivo Específico

**Crear un modelo que pueda predecir si una persona fallecida era HOMBRE o MUJER** basándose únicamente en:

1. **Su edad al morir**
2. **Cuándo murió** (fecha/temporada)
3. **Dónde murió** (región)

---

## 📈 Métricas de Éxito

### **Objetivos de Rendimiento:**
- **Accuracy > 80%**: El modelo predice correctamente el sexo en 8 de cada 10 casos
- **F1-Score > 75%**: Buen balance entre precisión y recall
- **Balanced Accuracy**: Que funcione bien para ambos géneros
- **ROC-AUC > 0.8**: Excelente capacidad discriminativa

### **Métricas Adicionales:**
- **Precision por clase**: Análisis detallado por género
- **Recall por clase**: Sensibilidad por género
- **Cross-Validation**: Validación robusta con k=5

## 🛠️ Metodología

### **1. Preparación de Datos:**
- **Muestra estratificada**: 100,000 registros manteniendo proporción de clases

### **2. Feature Engineering:**
- **Features temporales cíclicos**: mes_sin, mes_cos, etc.

- **Features de interacción**: edad × temporada, edad × región

### **3. Modelos a Evaluar:**
- **Logistic Regression**: Baseline con regularización
- **Random Forest**: Ensemble con múltiples árboles
- **Gradient Boosting**: Boosting adaptativo
- **Support Vector Machine**: Máquinas de soporte vectorial
- **K-Nearest Neighbors**: Vecinos más cercanos
- **Decision Tree**: Árbol de decisión

### **4. Optimización:**
- **GridSearchCV**: Búsqueda exhaustiva de hiperparámetros
- **Cross-Validation**: k=5 folds para validación robusta
- **Normalización**: StandardScaler para features numéricos

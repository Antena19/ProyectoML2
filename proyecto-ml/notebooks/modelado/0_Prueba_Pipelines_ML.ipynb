````python
# Setup inicial
import pandas as pd
import numpy as np
from pathlib import Path

# Configurar Kedro
from kedro.framework.session import KedroSession

project_path = Path.cwd().parent.parent
session = KedroSession.create(project_path=project_path)
context = session.load_context()
catalog = context.catalog

print("✅ Sesión Kedro creada correctamente")
print(f"📁 Ruta del proyecto: {project_path}")
````

## 1. Verificar que los datos de entrada existen


````python
# Verificar que existen los datos necesarios
try:
    datos = catalog.load("datasets_finales_modelado")
    print("✅ datasets_finales_modelado cargado correctamente")
    print(f"   Tipo: {type(datos)}")
    if isinstance(datos, dict):
        print(f"   Keys disponibles: {list(datos.keys())}")
        for key, value in datos.items():
            if hasattr(value, 'shape'):
                print(f"   - {key}: {value.shape}")
    elif hasattr(datos, 'shape'):
        print(f"   Shape: {datos.shape}")
        print(f"   Columnas: {list(datos.columns)[:10]}...")
except Exception as e:
    print(f"❌ Error al cargar datos: {e}")
````

## 2. OPCIÓN A: Prueba Rápida (Solo preparación de datos)


````python
# Prueba rápida: Solo preparar datos (sin entrenar modelos)
# Esto toma solo ~5 segundos

from proyecto_ml.pipelines.clasificacion.nodos import preparar_datos_clasificacion

# Cargar parámetros
params = context.params

print("🧪 Probando preparación de datos para clasificación...")
try:
    datos_clasificacion = preparar_datos_clasificacion(datos, params.get('modelado', {}))
    print("✅ Preparación de datos exitosa!")
    print(f"   Problemas preparados: {list(datos_clasificacion.keys())}")
    
    for problema, datos_prep in datos_clasificacion.items():
        print(f"\n   📊 {problema.upper()}:")
        print(f"      - X_train: {datos_prep['X_train'].shape}")
        print(f"      - X_test: {datos_prep['X_test'].shape}")
        print(f"      - Features: {len(datos_prep['feature_names'])}")
        print(f"      - Clases: {datos_prep['n_clases']}")
        
except Exception as e:
    print(f"❌ Error: {e}")
    import traceback
    traceback.print_exc()
````

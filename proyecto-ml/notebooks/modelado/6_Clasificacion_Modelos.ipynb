## 3. Visualizaciones Comparati

```python
# Gráfico comparativo de Accuracy
tabla_comparativa['Accuracy_Val'] = tabla_comparativa['Accuracy (Test)'].astype(float)

fig, ax = plt.subplots(figsize=(14, 6))
problemas = tabla_comparativa['Problema'].unique()

for i, problema in enumerate(problemas):
    datos = tabla_comparativa[tabla_comparativa['Problema'] == problema]
    x = np.arange(len(datos))
    ax.bar(x + i*0.35, datos['Accuracy_Val'], width=0.35, label=problema)

ax.set_xlabel('Modelo', fontweight='bold')
ax.set_ylabel('Accuracy', fontweight='bold')
ax.set_title('Comparación de Accuracy - Modelos de Clasificación', fontsize=14, fontweight='bold')
ax.set_xticks(np.arange(6))
ax.set_xticklabels(datos['Modelo'], rotation=45, ha='right')
ax.legend()
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('comparacion_accuracy_clasificacion.png', dpi=300, bbox_inches='tight')
plt.show()

print("✓ Gráfico guardado")
```
